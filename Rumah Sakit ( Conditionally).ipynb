```python
print("=" * 72)
print("     SISTEM RUMAH SAKIT: EVALUASI MEDIS & ADMINISTRASI OPNAME     ")
print("=" * 72)

nama_pasien   = input("Masukkan Nama Pasien      : ")
usia_pasien   = int(input("Masukkan Usia Pasien      : "))

print("\n--- DAFTAR KELUHAN / KONDISI MEDIS OPNAME (PILIH 1-15) ---")
print(" 1. Gagal Jantung / Serangan Jantung         (Masuk: ICCU)")
print(" 2. Bayi Lahir Prematur / Komplikasi         (Masuk: NICU)")
print(" 3. Anak / Remaja Kondisi Kritis             (Masuk: PICU)")
print(" 4. Kecelakaan / Tabrakan / Jatuh dari Pohon (Masuk: ICU)")
print(" 5. Luka Tusuk / Operasi Bedah Darurat       (Masuk: ICU)")
print(" 6. Stroke Akut / Pendarahan Otak            (Masuk: HCU)")
print(" 7. Asma Berat / Sesak Napas Akut            (Masuk: HCU)")
print(" 8. Gagal Ginjal / Cuci Darah Darurat        (Masuk: HCU)")
print(" 9. Pasca Operasi Besar (Usus Buntu/Sesar)   (Masuk: PACU)")
print("10. Demam Berdarah (DBD) Kondisi Drop        (Masuk: Rawat Inap Reguler)")
print("11. Tipes / Typhus Parah                     (Masuk: Rawat Inap Reguler)")
print("12. Pneumonia / Infeksi Paru-paru            (Masuk: Rawat Inap Reguler)")
print("13. Patah Tulang / Fraktur Tulang            (Masuk: Rawat Inap Reguler)")
print("14. Keracunan Makanan Berat                  (Masuk: Rawat Inap Reguler)")
print("15. Diabetes Komplikasi / Luka Bernanah      (Masuk: Rawat Inap Reguler)")
print("-------------------------------------------------------------------")

pilihan_keluhan = int(input("Pilih Nomor Keluhan Penyakit (1-15) : "))

print("\n" + "=" * 72)
print(f"HASIL ANALISIS MEDIS & PENEMPATAN RUANGAN: {nama_pasien.upper()}")
print("=" * 72)
print("Nama Pasien        :", nama_pasien)
print("Usia Pasien        :", usia_pasien, "tahun")

nama_ruangan   = ""
tarif_per_hari = 0
catatan_dokter = ""

if pilihan_keluhan == 1:
    nama_ruangan   = "ICCU (Intensive Coronary Care Unit)"
    tarif_per_hari = 1500000
    catatan_dokter = "Pasien gangguan jantung. Tindakan: Pemantauan EKG 24 jam & obat pengencer darah."
elif pilihan_keluhan == 2:
    nama_ruangan   = "NICU (Neonatal Intensive Care Unit)"
    tarif_per_hari = 1300000
    catatan_dokter = "Pasien bayi baru lahir. Tindakan: Inkubator khusus dan bantuan oksigen intensif."
elif pilihan_keluhan == 3:
    nama_ruangan   = "PICU (Pediatric Intensive Care Unit)"
    tarif_per_hari = 1200000
    catatan_dokter = "Pasien anak kritis. Tindakan: Perawatan intensif pediatrik khusus."
elif pilihan_keluhan == 4:
    nama_ruangan   = "ICU (Intensive Care Unit)"
    tarif_per_hari = 1400000
    catatan_dokter = "Kasus darurat parah (kecelakaan/jatuh). Tindakan: Ventilator & pemantauan tanda vital."
elif pilihan_keluhan == 5:
    nama_ruangan   = "ICU (Intensive Care Unit)"
    tarif_per_hari = 1400000
    catatan_dokter = "Kasus bedah darurat/luka tusuk. Tindakan: Sterilisasi luka & stabilisasi kondisi."
elif pilihan_keluhan == 6:
    nama_ruangan   = "HCU (High Care Unit)"
    tarif_per_hari = 900000
    catatan_dokter = "Pasien stroke akut. Tindakan: Observasi kesadaran & fisioterapi awal."
elif pilihan_keluhan == 7:
    nama_ruangan   = "HCU (High Care Unit)"
    tarif_per_hari = 900000
    catatan_dokter = "Pasien asma berat. Tindakan: Nebulizer berkala & oksigen tekanan tinggi."
elif pilihan_keluhan == 8:
    nama_ruangan   = "HCU (High Care Unit)"
    tarif_per_hari = 900000
    catatan_dokter = "Pasien gagal ginjal. Tindakan: Persiapan hemodialisis (cuci darah) rutin."
elif pilihan_keluhan == 9:
    nama_ruangan   = "PACU (Post-Anesthesia Care Unit / Ruang Pulih)"
    tarif_per_hari = 1000000
    catatan_dokter = "Pasien pasca operasi. Tindakan: Pemantauan efek bius & pereda nyeri intensif."
elif pilihan_keluhan == 10:
    nama_ruangan   = "Rawat Inap Reguler (Bangsal / Ward)"
    tarif_per_hari = 350000
    catatan_dokter = "Pasien DBD drop. Tindakan: Infus cairan elektrolit & cek darah lengkap harian."
elif pilihan_keluhan == 11:
    nama_ruangan   = "Rawat Inap Reguler (Bangsal / Ward)"
    tarif_per_hari = 350000
    catatan_dokter = "Pasien tipes. Tindakan: Diet bubur lunak, antibiotik, & istirahat total."
elif pilihan_keluhan == 12:
    nama_ruangan   = "Rawat Inap Reguler (Bangsal / Ward)"
    tarif_per_hari = 400000
    catatan_dokter = "Pasien pneumonia. Tindakan: Uap/nebulizer & antibiotik per infus."
elif pilihan_keluhan == 13:
    nama_ruangan   = "Rawat Inap Reguler (Bangsal / Ward)"
    tarif_per_hari = 350000
    catatan_dokter = "Pasien patah tulang. Tindakan: Pemasangan bidai/gips & obat pereda nyeri."
elif pilihan_keluhan == 14:
    nama_ruangan   = "Rawat Inap Reguler (Bangsal / Ward)"
    tarif_per_hari = 300000
    catatan_dokter = "Pasien keracunan. Tindakan: Pemasangan infus pencuci lambung & obat lambung."
elif pilihan_keluhan == 15:
    nama_ruangan   = "Rawat Inap Reguler (Bangsal / Ward)"
    tarif_per_hari = 400000
    catatan_dokter = "Pasien diabetes komplikasi. Tindakan: Cek gula darah berkala & perawatan luka."
else:
    nama_ruangan   = "Tidak Valid"
    tarif_per_hari = 0
    catatan_dokter = "Nomor keluhan tidak terdaftar di sistem rumah sakit."

if nama_ruangan == "Tidak Valid":
    print("[ERROR] Mohon masukkan nomor pilihan yang benar antara 1 sampai 15.")
else:
    print("Ruangan Tujuan     :", nama_ruangan)
    print("Catatan Dokter     :", catatan_dokter)

    print("\n--- ANALISIS DOSIS & PENANGANAN BERDASARKAN USIA ---")
    if usia_pasien <= 5:
        kategori_usia = "Balita (Usia 0-5 tahun)"
        aturan_usia   = "Dosis obat cair/sirup khusus anak, wajib didampingi orang tua 24 jam di bangsal."
    elif usia_pasien >= 6 and usia_pasien <= 12:
        kategori_usia = "Anak-anak (Usia 6-12 tahun)"
        aturan_usia   = "Dosis obat disesuaikan berat badan anak, perhatian khusus pada kenyamanan psikologis."
    elif usia_pasien >= 13 and usia_pasien <= 19:
        kategori_usia = "Remaja (Usia 13-19 tahun)"
        aturan_usia   = "Dosis obat standar remaja, pengawasan nutrisi dan jam istirahat ketat."
    elif usia_pasien >= 20 and usia_pasien <= 59:
        kategori_usia = "Dewasa (Usia 20-59 tahun)"
        aturan_usia   = "Dosis obat normal penuh dewasa, mobilisasi mandiri bertahap."
    else: # usia_pasien >= 60
        kategori_usia = "Lansia / Lanjut Usia (Usia >= 60 tahun)"
        aturan_usia   = "Perhatian Khusus Lansia: Pengurangan dosis obat (mengingat fungsi organ), penjagaan risiko jatuh, & visit geriatri harian."

    print("Kategori Usia      :", kategori_usia)
    print("Aturan / Penanganan:", aturan_usia)
    print("-" * 72)

    lama_hari_awal = int(input("Masukkan Rencana Awal Dirawat Berapa Hari : "))

    tambahan_hari = 0
    if lama_hari_awal < 3:
        print("\n[Evaluasi Dokter]  : Rencana rawat singkat (< 3 hari). Belum pulih sepenuhnya.")
        tambahan_hari = 2
        print("Saran Dokter       : Masa rawat ditambah otomatis 2 hari.")
    elif lama_hari_awal >= 3 and lama_hari_awal < 5:
        print("\n[Evaluasi Dokter]  : Kondisi berangsur membaik, butuh observasi tambahan.")
        tambahan_hari = 1
        print("Saran Dokter       : Masa rawat ditambah otomatis 1 hari.")
    else:
        print("\n[Evaluasi Dokter]  : Rencana rawat memadai. Target penyembuhan tercapai.")
        tambahan_hari = 0
        print("Saran Dokter       : Tidak ada tambahan hari, pasien siap pulang.")

    total_hari  = lama_hari_awal + tambahan_hari
    biaya_kamar = total_hari * tarif_per_hari

    if total_hari >= 7:
        fasilitas_ketentuan = "Fasilitas VIP: Visite dokter spesialis senior, perawat khusus 1:1, & fisioterapi gratis."
        biaya_tindakan      = 1500000
    elif total_hari >= 3 and total_hari < 7:
        fasilitas_ketentuan = "Fasilitas Menengah: Konsultasi harian dokter spesialis, perawat jaga, & cek lab rutin."
        biaya_tindakan      = 600000
    else:
        fasilitas_ketentuan = "Fasilitas Standar: Kamar perawatan umum, obat oral/infus dasar, & makan 3x sehari."
        biaya_tindakan      = 200000

    print("Fasilitas Didapat  :", fasilitas_ketentuan)

    total_tagihan = biaya_kamar + biaya_tindakan

    print("\n--- RINCIAN ADMINISTRASI AKHIR ---")
    print("Rencana Hari Awal   :", lama_hari_awal, "hari")
    print("Tambahan Hari Dokter:", tambahan_hari, "hari (Disarankan)")
    print("Total Hari Dirawat  :", total_hari, "hari")
    print("Tarif Kamar         : Rp", f"{tarif_per_hari:,}".replace(",", "."))
    print("Biaya Kamar Total   : Rp", f"{biaya_kamar:,}".replace(",", "."))
    print("Biaya Tindakan/Medis: Rp", f"{biaya_tindakan:,}".replace(",", "."))
    print("------------------------------------------------------------------------")
    print("TOTAL TAGIHAN      : Rp", f"{int(total_tagihan):,}".replace(",", "."))

print("=" * 72)
print("        RUMAH SAKIT SEHAT SEJAHTERA - MELAYANI DENGAN SEPENUH HATI       ")
print("=" * 72)
```
